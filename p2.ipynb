```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = "The students are studying NLP, and they were running quickly!"

# 1. Lowercase
text = text.lower()
print("Lowercase:", text)

# 2. Remove punctuation
text = re.sub(r'[^\w\s]', '', text)
print("Cleaned:", text)

# 3. Tokenization
tokens = word_tokenize(text)
print("Tokens:", tokens)

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered = [w for w in tokens if w not in stop_words]
print("Without stopwords:", filtered)

# 5. Stemming
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in filtered]
print("Stems:", stems)

# 6. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(w) for w in filtered]
print("Lemmas:", lemmas)
```

```python
from nltk.util import ngrams
from collections import Counter

sentence = "i love natural language processing"
words = sentence.split()

# Unigrams
unigrams = Counter(ngrams(words, 1))
print("Unigrams:", unigrams)

# Bigrams
bigrams = Counter(ngrams(words, 2))
print("Bigrams:", bigrams)

# Trigrams
trigrams = Counter(ngrams(words, 3))
print("Trigrams:", trigrams)

# Bigram probability
bigram = ("love", "natural")
prob = bigrams[bigram] / unigrams[(bigram[0],)]

print("P(natural | love) =", prob)
```

```python
from collections import Counter

sentence = "i love nlp i love python"
words = sentence.split()

bigrams = Counter(zip(words[:-1], words[1:]))
unigrams = Counter(words)
V = len(set(words))

previous = "love"
next_word = "python"

count_bigram = bigrams[(previous, next_word)]
count_previous = unigrams[previous]

prob = (count_bigram + 1) / (count_previous + V)

print("Bigram count:", count_bigram)
print("Previous word count:", count_previous)
print("Vocabulary size:", V)
print("Smoothed probability:", prob)
```

```python
from collections import Counter

words = ["low", "lower", "lowest", "low"]

# Represent each word as characters plus an end marker
vocab = {
    tuple(list(word) + ["</w>"]): count
    for word, count in Counter(words).items()
}

def get_pairs(vocab):
    pairs = Counter()
    for word, count in vocab.items():
        for i in range(len(word) - 1):
            pairs[(word[i], word[i + 1])] += count
    return pairs

def merge_pair(pair, vocab):
    new_vocab = {}
    bigram = " ".join(pair)

    for word, count in vocab.items():
        word_string = " ".join(word)
        merged = word_string.replace(bigram, "".join(pair))
        new_vocab[tuple(merged.split())] = count

    return new_vocab

for step in range(5):
    pairs = get_pairs(vocab)

    if not pairs:
        break

    best_pair = pairs.most_common(1)[0][0]
    vocab = merge_pair(best_pair, vocab)

    print("Merge", best_pair, "->", vocab)
```

```python
import torch
import torch.nn as nn

# Training data
text = "i love nlp i love python"
words = text.split()

vocab = sorted(set(words))
word_to_id = {w: i for i, w in enumerate(vocab)}
id_to_word = {i: w for w, i in word_to_id.items()}

# Input: current word, target: next word
X = torch.tensor([word_to_id[w] for w in words[:-1]])
Y = torch.tensor([word_to_id[w] for w in words[1:]])

# Neural language model
class NeuralLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, 8)
        self.fc = nn.Linear(8, vocab_size)

    def forward(self, x):
        x = self.embedding(x)
        return self.fc(x)

model = NeuralLM(len(vocab))
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.05)

# Training
for epoch in range(200):
    output = model(X)
    loss = loss_fn(output, Y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("Training completed!")

# Predict next word
input_word = "love"
x = torch.tensor([word_to_id[input_word]])

with torch.no_grad():
    prediction = model(x).argmax(dim=1).item()

print("Input word:", input_word)
print("Predicted next word:", id_to_word[prediction])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

texts = [
    "I love this movie",
    "This movie is amazing",
    "I hate this movie",
    "This movie is terrible",
    "The film is wonderful",
    "The film is boring"
]

labels = [
    "positive", "positive",
    "negative", "negative",
    "positive", "negative"
]

model = make_pipeline(
    CountVectorizer(),
    MultinomialNB()
)

model.fit(texts, labels)

test_text = ["The movie is wonderful"]
prediction = model.predict(test_text)

print("Text:", test_text[0])
print("Sentiment:", prediction[0])
```

```python
import torch
import torch.nn as nn

# Three sequences, each with four numerical features
X = torch.tensor([
    [[1.0], [2.0], [3.0], [4.0]],
    [[4.0], [3.0], [2.0], [1.0]],
    [[1.0], [1.0], [2.0], [2.0]]
])

# Class labels: 0 or 1
Y = torch.tensor([0, 1, 0])

class RNNClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.rnn = nn.RNN(
            input_size=1,
            hidden_size=8,
            batch_first=True
        )
        self.fc = nn.Linear(8, 2)

    def forward(self, x):
        output, hidden = self.rnn(x)
        last_output = output[:, -1, :]
        return self.fc(last_output)

model = RNNClassifier()
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# Train
for epoch in range(100):
    predictions = model(X)
    loss = criterion(predictions, Y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("Training completed!")

# Predict
with torch.no_grad():
    output = model(X)
    predicted_classes = output.argmax(dim=1)

print("Actual:", Y.tolist())
print("Predicted:", predicted_classes.tolist())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

actual = ["positive", "negative", "positive",
          "negative", "positive"]

predicted = ["positive", "negative", "negative",
             "negative", "positive"]

print("Accuracy:", accuracy_score(actual, predicted))

print("Precision:",
      precision_score(actual, predicted,
                      pos_label="positive"))

print("Recall:",
      recall_score(actual, predicted,
                   pos_label="positive"))

print("F1-score:",
      f1_score(actual, predicted,
               pos_label="positive"))

print("\nConfusion matrix:")
print(confusion_matrix(
    actual, predicted,
    labels=["positive", "negative"]
))

print("\nClassification report:")
print(classification_report(actual, predicted))
```

```python
import torch
import torch.nn.functional as F

# Three word representations, each with two features
Q = torch.tensor([[1.0, 0.0]])
K = torch.tensor([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
])

V = torch.tensor([
    [10.0, 0.0],
    [0.0, 10.0],
    [5.0, 5.0]
])

# 1. Calculate attention scores
scores = torch.matmul(Q, K.T)

# 2. Scale scores
scores = scores / (K.shape[1] ** 0.5)

# 3. Convert scores into weights
weights = F.softmax(scores, dim=-1)

# 4. Calculate weighted output
output = torch.matmul(weights, V)

print("Scores:", scores)
print("Attention weights:", weights)
print("Attention output:", output)
```
